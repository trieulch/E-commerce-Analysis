```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv(r"D:\Python\page_performance_week_22_35.csv")
df = pd.DataFrame(data)
# Sessions
plt.figure()
plt.plot(data["year_week"], data["total_sessions_homepage"], label="homepage")
plt.plot(data["year_week"], data["total_sessions_lander_1"], label="lander-1")
plt.xlabel("year_week")
plt.ylabel("sessions")
plt.title("Sessions by Page")
plt.legend()
plt.show()

# Bounced sessions
plt.figure()
plt.plot(data["year_week"], data["total_bounced_sessions_home"], label="homepage")
plt.plot(data["year_week"], data["total_bounced_sessions_lander_1"], label="lander-1")
plt.xlabel("year_week")
plt.ylabel("bounced_sessions")
plt.title("Bounced Sessions by Page")
plt.legend()
plt.show()

# Bounce ratio
plt.figure()
plt.plot(data["year_week"], data["bounced_homepage_ratio"], label="homepage")
plt.plot(data["year_week"], data["bounced_lander_1_ratio"], label="lander-1")
plt.xlabel("year_week")
plt.ylabel("bounce_ratio")
plt.title("Bounce Ratio by Page")
plt.legend()
plt.show()
```
